# Train-only exploratory analysis
This educational companion visualizes the **training partition only**. First download UCI HAR and install the package. Production execution uses the CLI (`scripts/inspect_dataset.py`).

```python
from pathlib import Path
from sensor_classifier.config import load_config
from sensor_classifier.data.loading import load_dataset
from sensor_classifier.analysis.data_quality import quality_report
from sensor_classifier.analysis.exploratory import explore
root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
settings = load_config(root)
dataset = load_dataset(settings)
quality = quality_report(dataset.X_train, dataset.y_train, dataset.labels)
{k: quality[k] for k in ('rows','features','classes','missing_cells','duplicate_predictor_rows','label_counts')}
```

```python
output = root / 'results' / 'notebook-eda' / 'figures'
explore(dataset.X_train, dataset.y_train, output)
print('Train-only figures:', output)
```
